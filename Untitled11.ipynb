```python
# ==========================================================
# Symptom Information Bot (LLM Mini Project - PS 56)
# Client / Student : Viraj Sukhdev Kalbhor
# Roll No: 56 | Enrollment ID: ADT23SOCB1318
# ==========================================================
import gradio as gr
from google import genai
from google.genai import types

CLIENT_NAME "Viraj Sukhdev Kalbhor"

client = genai.Client(api_key=API_KEY)


EMERGENCY_BANNER = ("🚨 **This may be a medical emergency.** Call **112** or **108** "
                    "(ambulance) or go to the nearest hospital immediately.\n\n")

EMERGENCY_WORDS = ["chest pain", "heart attack", "can't breathe", "cannot breathe",
                   "difficulty breathing", "shortness of breath", "severe bleeding",
                   "unconscious", "fainted", "stroke", "slurred speech", "seizure",
                   "suicide", "kill myself", "overdose", "poison", "coughing blood",
                   "vomiting blood", "anaphylaxis"]

SYSTEM_PROMPT = """You are "Symptom Information Bot", a careful health-information assistant.
RULES:
1. Give only GENERAL educational information. NEVER diagnose ("you have X" is forbidden).
2. NEVER prescribe medicines, doses or treatments.
3. Use the user's age and duration to make the answer more relevant.
4. ALWAYS suggest the TYPE of doctor to visit. Start with General Physician if unsure.
5. If symptoms may be serious, begin with "EMERGENCY:" and say to call 112/108.
6. If the message is not about health, say you only help with symptom information.
7. Simple, kind language. Under 250 words.
FORMAT:
**Urgency level:** Self-care / See a doctor soon / Emergency
**About this symptom:** ...
**Possible general causes:** ...
**Self-care (mild cases):** ...
**Warning signs - see a doctor soon if:** ...
**Doctor to visit:** ...
**What the doctor may ask you:** ...
**Tests a doctor may suggest:** ...
**How to prepare for the visit:** ...
"""

def symptom_bot(symptom, age, duration):
    if not symptom.strip():
        return "Please type a symptom."
    banner = EMERGENCY_BANNER if any(w in symptom.lower() for w in EMERGENCY_WORDS) else ""
    prompt = f"Symptom: {symptom}\nAge: {age}\nDuration: {duration}"
    try:
        r = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=prompt,
            config=types.GenerateContentConfig(
                system_instruction=SYSTEM_PROMPT, temperature=0.3, max_output_tokens=2000))
        answer = r.text
    except Exception as e:
        answer = f"Error: {e}"
    footer = f"\n\n---\n{DISCLAIMER}\n\n*Project by {CLIENT_NAME} | Roll No. 56 | ADT23SOCB1318*"
    return banner + answer + footer

demo = gr.Interface(
    fn=symptom_bot,
    inputs=[
        gr.Textbox(label="Describe your symptom", placeholder="e.g., sore throat and cough"),
        gr.Number(label="Age", value=20),
        gr.Textbox(label="How long?", placeholder="e.g., 3 days"),
    ],
    outputs=gr.Markdown(label="Answer"),
    title=f"🩺 Symptom Information Bot - by {CLIENT_NAME}",
    description=DISCLAIMER,
    examples=[["Headache and nausea", 20, "since morning"],
              ["Red itchy rash on my arm", 22, "2 days"],
              ["Chest pain and sweating", 45, "30 minutes"],
              ["Which medicine should I take for fever?", 20, "1 day"]],
)
demo.launch()
```
